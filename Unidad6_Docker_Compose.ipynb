## Introducción al Contenido Avanzado

En la unidad principal aprendimos a empaquetar nuestro microservicio FastAPI en un contenedor individual con Docker. Pero una solución digital real no consta de un solo servicio: necesita conectarse a una base de datos relacional (como PostgreSQL) y a veces a un motor de caché (como Redis).

Instalar todo esto de forma local en tu computadora puede causar problemas de compatibilidad y puertos. **Docker Compose** es la herramienta que nos permite definir y correr aplicaciones multi-contenedor en un único comando utilizando un archivo de configuración declarativo llamado `docker-compose.yml`.


## 1. Estructura del Archivo `docker-compose.yml`

El archivo `docker-compose.yml` especifica qué servicios se deben crear, sus puertos, volúmenes de almacenamiento y variables de entorno para vincularlos entre sí.

### Ejemplo de Configuración: App + Base de Datos PostgreSQL

A continuación se presenta un archivo de configuración real de Docker Compose para una arquitectura FastAPI + PostgreSQL:


```python
# Escribimos el docker-compose.yml en el entorno local
compose_content = """
version: '3.8'

services:
  # Servicio 1: Nuestra base de datos relacional PostgreSQL
  db:
    image: postgres:15-alpine
    container_name: fintech_postgres_db
    environment:
      POSTGRES_USER: admin_user
      POSTGRES_PASSWORD: super_password_123
      POSTGRES_DB: fintech_db
    ports:
      - "5432:5432"
    volumes:
      - postgres_data:/var/lib/postgresql/data

  # Servicio 2: Nuestro microservicio FastAPI
  web:
    build: .
    container_name: fastapi_app_container
    ports:
      - "8000:8000"
    environment:
      - DATABASE_URL=postgresql://admin_user:super_password_123@db:5432/fintech_db
    depends_on:
      - db # Asegura que la base de datos se inicie antes que la app web

volumes:
  postgres_data: # Permite persistir los datos de la base de datos fuera del ciclo de vida del contenedor
"""

with open("docker-compose.yml", "w") as f:
    f.write(compose_content.strip())
    
print("Archivo docker-compose.yml configurado con éxito.")
```

## 2. Comandos Clave de Orquestación

Cuando trabajes localmente desde la terminal de tu computadora con Docker Compose instalado, los comandos principales a ejecutar son:
- `docker-compose up --build`: Construye las imágenes y levanta todos los contenedores en segundo plano.
- `docker-compose down`: Apaga todos los contenedores y destruye la red virtual, liberando memoria y puertos.
- `docker-compose logs -f web`: Permite ver los logs en tiempo real de nuestra aplicación web FastAPI.
